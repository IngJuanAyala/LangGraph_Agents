```python
from langchain_core.messages import AIMessage

ai_msg = AIMessage(content="Hello, I am an AI message.")
print(ai_msg.text)  # Output: Hello, I am an AI message.
```

```python
from langchain_core.messages import HumanMessage

human_msg = HumanMessage(content="Hello, I am an human message.")
print(human_msg.text)  # Output: Hello, I am an AI message.
```

```python
history = [human_msg, ai_msg]
for msg in history:
    msg.pretty_print()
```
