```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(model="claude-sonnet-5")
response = llm.invoke("Hola soy Juan y mi telefono es 3333333333") 
response.pretty_print()
```

```python
from pydantic import BaseModel, Field

class ContactInfo(BaseModel):
    name: str = Field(description="The name of the contact")
    phone_number: str = Field(description="The phone number of the contact")
    email: str = Field(description="The email address of the contact", default=None)
    tone: str = Field(description="The tone of the message", default="neutral")
    sentiment: str = Field(description="The sentiment of the message", default="neutral")
    age: int = Field(description="The age of the contact", default=None)

llm_with_structured_output = llm.with_structured_output(schema=ContactInfo)

messages = [
    {"role": "system", "content": "You are a helpful assistant that extracts contact information from a given conversation."},
    {"role": "user", "content": "Hola soy Juan y mi telefono es 3333333333 mi correo es correo@correo.com"}
]

response = llm_with_structured_output.invoke(messages)
print(response)
```
