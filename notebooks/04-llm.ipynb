```python
from langchain_openai import ChatOpenAI

gpt4o_llm = ChatOpenAI(model_name="gpt-4o", temperature=0)
response = gpt4o_llm.invoke("Hello, how are you?") 
response.pretty_print()

```

```python
from langchain_openai import ChatOpenAI

gpt35_llm = ChatOpenAI(model_name="gpt-3.5-turbo-0125", temperature=0)
response = gpt35_llm.invoke("Hello, how are you?") 
response.pretty_print()
```

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage

msg_1 = SystemMessage(content="You are a helpful assistant that talks in spanish")
msg_2 = HumanMessage(content="Me llamo Juan")
msg_3 = AIMessage(content="Hola Juan, como estas?")
msg_4 = HumanMessage(content="Como me llamo?")
history = [msg_1, msg_2, msg_3, msg_4]
response = gpt4o_llm.invoke(history)
response.pretty_print()
```

```python
from langchain_anthropic import ChatAnthropic

claude_llm = ChatAnthropic(model="claude-sonnet-5")
response = claude_llm.invoke(history)
response.pretty_print()
```
